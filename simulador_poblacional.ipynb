## 1. Proyección por región con la banda de escenarios
Se elige la región, el método y el paso. La línea negra es el simulador, la gris gruesa es la integral exacta y la banda azul es el rango de los escenarios de la ONU.

```python
from ipywidgets import interact, Dropdown, FloatSlider, IntSlider
import interactivo as it, simulador as sim
regiones = sorted(set(sim.CLAVES_TASA.values()))
interact(it.grafica_region,
         region=Dropdown(options=[(sim.NOMBRES[c], c) for c in regiones], value='COL', description='Región'),
         metodo=Dropdown(options=['rk4', 'euler'], description='Método'),
         paso=Dropdown(options=[1.0, 2.0, 5.0, 10.0], description='Paso (años)'),
         hasta=IntSlider(min=2030, max=2100, step=10, value=2100, description='Hasta'));
```

## 2. Inercia demográfica
Con β = γ (reemplazo) la población sigue creciendo hasta unos 9.123 millones. Se puede mover la natalidad, la inmigración y la vida media después de los 50.

```python
interact(it.grafica_inercia,
         beta_sobre_gamma=FloatSlider(min=0.8, max=1.2, step=0.05, value=1.0, description='β / γ'),
         inmigracion=FloatSlider(min=0, max=5, step=0.5, value=0, description='m (M/año)'),
         delta_anios=IntSlider(min=15, max=35, value=27, description='Años >50'));
```
